```python
import numpy as np

# Activation Function
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

# Derivative
def sigmoid_derivative(x):
    return x * (1 - x)

# Dataset: Study Hours → Pass(1)/Fail(0)
# (You can explain this easily in viva)
X_original = np.array([[1],
                       [2],
                       [3],
                       [4],
                       [5],
                       [6],
                       [7],
                       [8]])

y = np.array([[0],
              [0],
              [0],
              [0],
              [1],
              [1],
              [1],
              [1]])

# Normalize input
max_val = np.max(X_original)
X = X_original / max_val

# Network Structure
input_neurons = 1
hidden_neurons = 3
output_neurons = 1

# Initialize weights & bias
np.random.seed(1)
wh = np.random.rand(input_neurons, hidden_neurons)
bh = np.random.rand(1, hidden_neurons)
wo = np.random.rand(hidden_neurons, output_neurons)
bo = np.random.rand(1, output_neurons)

learning_rate = 0.1
epochs = 20   # Keep small to display steps

print("----- TRAINING STARTED -----\n")

for epoch in range(epochs):

    # Forward Propagation
    hidden = sigmoid(np.dot(X, wh) + bh)
    output = sigmoid(np.dot(hidden, wo) + bo)

    # Error
    error = y - output

    # Backpropagation
    d_output = error * sigmoid_derivative(output)
    d_hidden = d_output.dot(wo.T) * sigmoid_derivative(hidden)

    # Update weights
    wo += hidden.T.dot(d_output) * learning_rate
    bo += np.sum(d_output, axis=0)
    wh += X.T.dot(d_hidden) * learning_rate
    bh += np.sum(d_hidden, axis=0)

    # Display each epoch
    print(f"Epoch {epoch+1}")
    print("Predicted Output:\n", output)
    print("Error:\n", error)
    print("Mean Error:", np.mean(np.abs(error)))
    print("-"*40)

print("\n----- TRAINING COMPLETED -----\n")
print("Final Output:\n", output)

# -------------------------------
# User Input
# -------------------------------
print("\n----- TEST WITH USER INPUT -----")

hours = float(input("Enter study hours: "))

# Normalize input
user_input = np.array([[hours]]) / max_val

hidden_user = sigmoid(np.dot(user_input, wh) + bh)
output_user = sigmoid(np.dot(hidden_user, wo) + bo)

# Convert to binary result
result = 1 if output_user[0][0] >= 0.5 else 0

print("Predicted Value:", output_user[0][0])
print("Result:", "Pass" if result == 1 else "Fail")
```
