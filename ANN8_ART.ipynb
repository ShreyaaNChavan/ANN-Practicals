```python
import numpy as np

# -------------------------------
# STEP 1: Input Patterns
# -------------------------------
patterns = [
    np.array([1, 1, 0, 0]),
    np.array([1, 1, 0, 1]),
    np.array([0, 0, 1, 1])
]

# -------------------------------
# STEP 2: Parameters
# -------------------------------
vigilance = 0.6
epochs = 3   # Artificial epochs for demonstration

# -------------------------------
# STEP 3: Categories
# -------------------------------
categories = []

# -------------------------------
# STEP 4: Match Function
# -------------------------------
def match_score(X, W):
    intersection = np.sum(X * W)
    return intersection / np.sum(X)

# -------------------------------
# STEP 5: Training with Epochs
# -------------------------------
print("----- TRAINING STARTED -----\n")

for epoch in range(epochs):
    print(f"\n=== Epoch {epoch+1} ===")

    for i, X in enumerate(patterns):
        print(f"\nProcessing Pattern {i+1}: {X}")

        assigned = False

        for j, W in enumerate(categories):
            score = match_score(X, W)
            print(f"  Match with Category {j}: {score:.2f}")

            if score >= vigilance:
                print(f"  → Assigned to Category {j}")
                categories[j] = X * W   # Update
                assigned = True
                break

        if not assigned:
            print("  → New Category Created")
            categories.append(X.copy())

print("\n----- TRAINING COMPLETED -----")

# -------------------------------
# STEP 6: Final Categories
# -------------------------------
print("\nFinal Categories:")
for i, cat in enumerate(categories):
    print(f"Category {i}: {cat}")

# -------------------------------
# STEP 7: USER INPUT
# -------------------------------
print("\n----- TEST WITH USER INPUT -----")

user_pattern = list(map(int, input("Enter 4 binary values (space separated): ").split()))
user_pattern = np.array(user_pattern)

assigned = False

for j, W in enumerate(categories):
    score = match_score(user_pattern, W)
    print(f"Match with Category {j}: {score:.2f}")

    if score >= vigilance:
        print(f"→ User pattern belongs to Category {j}")
        assigned = True
        break

if not assigned:
    print("→ User pattern forms a NEW category")
```
